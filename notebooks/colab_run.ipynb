## Gemini 실행

Colab 왼쪽 열쇠 아이콘(보안 비밀)에 `GEMINI_API_KEY`를 추가하거나, 레포 루트 `.env`에 넣으세요. 키는 출력하지 않습니다.

```python
# Colab secrets -> environment (value is never printed)
import os
try:
    from google.colab import userdata
    os.environ.setdefault('GEMINI_API_KEY', userdata.get('GEMINI_API_KEY') or '')
except Exception:
    pass  # local kernel: kosis_chat reads .env
print('GEMINI_API_KEY set:', bool(os.environ.get('GEMINI_API_KEY')))
```
